```python
import os
# Install dependencies

# Import required libraries
from unsloth import FastLanguageModel
from transformers import TrainingArguments
from datasets import load_dataset,Dataset
import torch
import pandas as pd

# Load Llama-3 model and tokenizer
model, tokenizer = FastLanguageModel.from_pretrained(
    model_name="unsloth/llama-3-8b",
    max_seq_length=4096,
    dtype=torch.float16,  # Use torch.float16 instead of "float16"
    load_in_4bit=True,
)

# Load the CSV file into a pandas DataFrame
df = pd.read_csv('/content/dataset.csv')

# Convert the DataFrame to a Hugging Face Dataset
dataset = Dataset.from_pandas(df)

# Define training arguments
training_args = TrainingArguments(
    output_dir="./results",
    per_device_train_batch_size=4,
    num_train_epochs=3,
    save_steps=500,
    save_total_limit=2,
)

# Fine-tune the model
model.fit(
    dataset["train"],
    tokenizer=tokenizer,
    training_args=training_args,
    lora_r=8,
    lora_alpha=16,
    lora_dropout=0.05,
)

# Save model for Ollama deployment
output_path = "./ollama_model"
model.save_to_ollama(path=output_path, model_name="llama-3-8b-custom")

print(f"Model saved at {output_path}")
```
