```python
# prompt: # 2. Import libraries
from unsloth import FastLanguageModel
from datasets import load_dataset
from transformers import TrainingArguments
import torch
import pandas as pd
 # 3. Load your dataset from CSV
df = pd.read_csv("/content/merged_dataset.csv")  # Replace with your file
df = df[['title','selftext', 'Label']]  # Optional: can include title or other fields
df = df.dropna()
 # 4. Preprocess for instruction-tuning style
def format_prompt(row):
  return {
    "input": f"User: {row['title']}{row['selftext']}\nSentiment: {row['Label']}\nAssistant:", # Changed this line
    "output": "I'm really sorry you're feeling this way. You're not alone. Let's talk more."
}
formatted_dataset = df.apply(format_prompt, axis=1).tolist()
dataset = pd.DataFrame(formatted_dataset)

# Assuming 'formatted_dataset' from the previous code block is available

dataset = pd.DataFrame(formatted_dataset)
dataset.to_csv('formatted_dataset.csv', index=False)
```

```python

# 5. Convert to HuggingFace Dataset
from datasets import Dataset
train_data = Dataset.from_pandas(dataset)

# 6. Load Mistral-7B with Unsloth
model, tokenizer = FastLanguageModel.from_pretrained(
    model_name="unsloth/mistral-7b-instruct-v0.2",
    max_seq_length=2048,
    dtype=None,
    load_in_4bit=True,
    token=True,
)

# 7. Add LoRA adapters
model = FastLanguageModel.get_peft_model(
    model,
    r=16,
    target_modules=["q_proj", "v_proj"],
    lora_alpha=16,
    lora_dropout=0.05,
    bias="none",
    use_gradient_checkpointing=True,
    random_state=42,
    use_rslora=False,
    loftq_config=None,
)

# 8. Tokenize
def format_and_tokenize(example):
    return tokenizer(f"<|system|>\nYou are a kind and empathetic mental health assistant.\n<|user|>\n{example['input']}\n<|assistant|>\n{example['output']}",
                     truncation=True, padding="max_length", max_length=2048)

train_data = train_data.map(format_and_tokenize)

# 9. Set training arguments
training_args = TrainingArguments(
    output_dir="./results",
    per_device_train_batch_size=2,
    gradient_accumulation_steps=8,
    num_train_epochs=1,
    save_steps=100,
    save_total_limit=2,
    logging_steps=10,
    learning_rate=2e-4,
    max_grad_norm=0.3,
    gradient_checkpointing=True,
    fp16=True,
    optim="paged_adamw_32bit",
    warmup_ratio=0.03,
    weight_decay=0.01,
    report_to="tensorboard",
    remove_unused_columns=False,
    prediction_loss_only=True
)

# 10. Train the model
FastLanguageModel.prepare_for_training(model)
model.train()
trainer = FastLanguageModel.get_trainer(model=model, tokenizer=tokenizer, args=training_args, train_dataset=train_data)
trainer.train()

# 11. Save model for use with Ollama or inference
model.save_pretrained("/content/empathetic-mistral")
tokenizer.save_pretrained("/content/empathetic-mistral")

```
